```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from torch.distributions.multivariate_normal import MultivariateNormal

# -----------------------------
# Device setup
# -----------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Optional: reproducibility
torch.manual_seed(0)


# -----------------------------
# MMD function
# -----------------------------

def MMD(x, y, kernel="multiscale"):
    """
    Empirical maximum mean discrepancy.

    Args:
        x: first sample, shape (n, d)
        y: second sample, shape (m, d)
        kernel: "multiscale" or "rbf"

    Returns:
        scalar MMD value
    """

    # Squared pairwise distances
    dxx = torch.cdist(x, x, p=2) ** 2
    dyy = torch.cdist(y, y, p=2) ** 2
    dxy = torch.cdist(x, y, p=2) ** 2

    XX = torch.zeros_like(dxx)
    YY = torch.zeros_like(dyy)
    XY = torch.zeros_like(dxy)

    if kernel == "multiscale":
        bandwidth_range = [0.2, 0.5, 0.9, 1.3]

        for a in bandwidth_range:
            XX += a**2 / (a**2 + dxx)
            YY += a**2 / (a**2 + dyy)
            XY += a**2 / (a**2 + dxy)

    elif kernel == "rbf":
        bandwidth_range = [10, 15, 20, 50]

        for a in bandwidth_range:
            XX += torch.exp(-0.5 * dxx / a)
            YY += torch.exp(-0.5 * dyy / a)
            XY += torch.exp(-0.5 * dxy / a)

    else:
        raise ValueError("kernel must be either 'multiscale' or 'rbf'")

    return XX.mean() + YY.mean() - 2 * XY.mean()


# -----------------------------
# Generate samples
# -----------------------------

m = 20

x_mean = torch.zeros(2, device=device) + 1
y_mean = torch.zeros(2, device=device)

x_cov = 2 * torch.eye(2, device=device)
y_cov = 3 * torch.eye(2, device=device) - 1

px = MultivariateNormal(x_mean, x_cov)
qy = MultivariateNormal(y_mean, y_cov)

x = px.sample((m,))
y = qy.sample((m,))

result = MMD(x, y, kernel="multiscale")

print(f"MMD result of X and Y is {result.item()}")


# -----------------------------
# Plotting setup
# -----------------------------

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4), dpi=100)

grid_size = 100

x1_val = np.linspace(-5, 5, num=grid_size)
x2_val = np.linspace(-5, 5, num=grid_size)

x1, x2 = np.meshgrid(x1_val, x2_val)

# Shape: (grid_size, grid_size, 2)
grid_np = np.stack([x1, x2], axis=-1)

# Convert plotting grid to Torch tensor
grid = torch.tensor(grid_np, dtype=torch.float32, device=device)

# Evaluate densities using PyTorch instead of scipy
px_grid = torch.exp(px.log_prob(grid)).detach().cpu().numpy()
qy_grid = torch.exp(qy.log_prob(grid)).detach().cpu().numpy()


# -----------------------------
# Plot X distribution
# -----------------------------

CS1 = ax1.contourf(x1, x2, px_grid, 100, cmap=plt.cm.YlGnBu)

ax1.set_title(r"Distribution of $X \sim P(X)$")
ax1.set_xlabel(r"$x_1$")
ax1.set_ylabel(r"$x_2$")
ax1.set_aspect("equal")

ax1.scatter(
    x[:10, 0].detach().cpu().numpy(),
    x[:10, 1].detach().cpu().numpy(),
    label=r"$X$ Samples",
    marker="o",
    facecolor="r",
    edgecolor="k"
)

ax1.legend()


# -----------------------------
# Plot Y distribution
# -----------------------------

CS2 = ax2.contourf(x1, x2, qy_grid, 100, cmap=plt.cm.YlGnBu)

ax2.set_title(r"Distribution of $Y \sim Q(Y)$")
ax2.set_xlabel(r"$y_1$")
ax2.set_ylabel(r"$y_2$")
ax2.set_aspect("equal")

ax2.scatter(
    y[:10, 0].detach().cpu().numpy(),
    y[:10, 1].detach().cpu().numpy(),
    label=r"$Y$ Samples",
    marker="o",
    facecolor="r",
    edgecolor="k"
)

ax2.legend()


# -----------------------------
# Colorbar and title
# -----------------------------

fig.subplots_adjust(right=0.8)

cbar_ax = fig.add_axes([0.85, 0.15, 0.02, 0.7])
cbar = fig.colorbar(CS2, cax=cbar_ax)
cbar.ax.set_ylabel("Density")

plt.suptitle(
    f"MMD result: {round(result.item(), 3)}",
    y=0.95,
    fontweight="bold"
)

plt.show()
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

from pathlib import Path
from torch.distributions.multivariate_normal import MultivariateNormal


# -----------------------------
# Font setup
# -----------------------------

font_folder = Path(r"C:\USERS\LEO\APPDATA\LOCAL\MICROSOFT\WINDOWS\FONTS")

for font_file in [
    "CMUNRM.TTF",  # Roman
    "CMUNTI.TTF",  # Italic
    "CMUNBX.TTF",  # Bold
    "CMUNBI.TTF",  # Bold Italic
]:
    fm.fontManager.addfont(str(font_folder / font_file))

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["CMU Serif"],

    "mathtext.fontset": "cm",
    "axes.unicode_minus": False,

    "font.size": 10,
    "axes.labelsize": 10,
    "axes.titlesize": 10,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 10,

    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})


# -----------------------------
# Device setup
# -----------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

torch.manual_seed(0)


# -----------------------------
# MMD function
# -----------------------------

def MMD(x, y, kernel="multiscale"):
    """
    Empirical maximum mean discrepancy.

    Args:
        x: first sample, shape (n, d)
        y: second sample, shape (m, d)
        kernel: "multiscale" or "rbf"

    Returns:
        scalar MMD value
    """

    dxx = torch.cdist(x, x, p=2) ** 2
    dyy = torch.cdist(y, y, p=2) ** 2
    dxy = torch.cdist(x, y, p=2) ** 2

    XX = torch.zeros_like(dxx)
    YY = torch.zeros_like(dyy)
    XY = torch.zeros_like(dxy)

    if kernel == "multiscale":
        bandwidth_range = [0.2, 0.5, 0.9, 1.3]

        for a in bandwidth_range:
            XX += a**2 / (a**2 + dxx)
            YY += a**2 / (a**2 + dyy)
            XY += a**2 / (a**2 + dxy)

    elif kernel == "rbf":
        bandwidth_range = [10, 15, 20, 50]

        for a in bandwidth_range:
            XX += torch.exp(-0.5 * dxx / a)
            YY += torch.exp(-0.5 * dyy / a)
            XY += torch.exp(-0.5 * dxy / a)

    else:
        raise ValueError("kernel must be either 'multiscale' or 'rbf'")

    return XX.mean() + YY.mean() - 2 * XY.mean()


# -----------------------------
# Generate samples
# -----------------------------

m = 20

x_mean = torch.zeros(2, device=device) + 1
y_mean = torch.zeros(2, device=device)

x_cov = 2 * torch.eye(2, device=device)
y_cov = 3 * torch.eye(2, device=device) - 1

px = MultivariateNormal(x_mean, x_cov)
qy = MultivariateNormal(y_mean, y_cov)

x = px.sample((m,))
y = qy.sample((m,))

result = MMD(x, y, kernel="multiscale")

print(f"MMD result of X and Y is {result.item():.4f}")


# -----------------------------
# Create density grid
# -----------------------------

grid_size = 100

x1_val = np.linspace(-5, 5, grid_size)
x2_val = np.linspace(-5, 5, grid_size)

x1, x2 = np.meshgrid(x1_val, x2_val)

grid_np = np.stack([x1, x2], axis=-1)
grid = torch.tensor(grid_np, dtype=torch.float32, device=device)

px_grid = torch.exp(px.log_prob(grid)).detach().cpu().numpy()
qy_grid = torch.exp(qy.log_prob(grid)).detach().cpu().numpy()


# -----------------------------
# Plot
# -----------------------------

cm = 1 / 2.54

fig, (ax1, ax2) = plt.subplots(
    1,
    2,
    figsize=(16 * cm, 6 * cm),
    dpi=150,
    layout="constrained"
)

levels = np.linspace(0, 0.1, 101)

CS1 = ax1.contourf(
    x1,
    x2,
    px_grid,
    levels=levels,
    cmap=plt.cm.YlGnBu,
    vmin=0,
    vmax=0.1
)

ax1.scatter(
    x[:10, 0].detach().cpu().numpy(),
    x[:10, 1].detach().cpu().numpy(),
    label=r"$X$ samples",
    marker="o",
    facecolor="r",
    edgecolor="k",
    s=25,
    linewidth=0.7
)

ax1.set_title(r"$X \sim P(X)$")
ax1.set_xlabel(r"$x_1$")
ax1.set_ylabel(r"$x_2$")
ax1.set_aspect("equal")
ax1.legend(frameon=False)


CS2 = ax2.contourf(
    x1,
    x2,
    qy_grid,
    levels=levels,
    cmap=plt.cm.YlGnBu,
    vmin=0,
    vmax=0.1
)

ax2.scatter(
    y[:10, 0].detach().cpu().numpy(),
    y[:10, 1].detach().cpu().numpy(),
    label=r"$Y$ samples",
    marker="o",
    facecolor="r",
    edgecolor="k",
    s=25,
    linewidth=0.7
)

ax2.set_title(r"$Y \sim Q(Y)$")
ax2.set_xlabel(r"$y_1$")
ax2.set_ylabel(r"$y_2$")
ax2.set_aspect("equal")
ax2.legend(frameon=False)


# -----------------------------
# Colorbar and title
# -----------------------------

cbar = fig.colorbar(
    CS2,
    ax=[ax1, ax2],
    shrink=1,
    pad=0.03,
    ticks=np.arange(0, 0.1001, 0.02)
)

cbar.ax.set_ylabel(r"Density")

#fig.suptitle(
#    rf"MMD result: {result.item():.3f}",
#    fontweight="bold"
#)


# -----------------------------
# Save and show
# -----------------------------

fig.savefig(
    "../1_latex/1.5_figures/test figures/mmd_plot.pdf",
    bbox_inches="tight"
)

plt.show()
```
